Polynomial regression
<a class="anchor" id="1"></a> 

```python
import pandas as pd

df = pd.read_csv('../data/sales.csv')

df.head()
```

```python
X= df["Temperature"]
y= df["Sales"]
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color = 'red')
plt.title('Ice Cream Selling Data')
plt.xlabel('Temperature')
plt.ylabel('Sales (units)')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()
linear_model.fit(df[["Temperature"]], y)

plt.scatter(X , y, color = 'red')
plt.plot(X, linear_model.predict(df[["Temperature"]]), color = 'blue')
plt.title('Linear Regression')
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly_reg = PolynomialFeatures(degree=2)
X_poly = poly_reg.fit_transform(df[["Temperature"]])

linear_model2 = LinearRegression()
linear_model2.fit(X_poly, y)

plt.scatter(X, y, color = 'red')
plt.plot(X, linear_model2.predict(X_poly), color = 'blue')
plt.title('Polynoamial Regression')
plt.show()
```

```python
# the intercept and coefficients of the polynomial model form this equation:
print('Intercept: ', linear_model2.intercept_)
print('Coefficients: ', linear_model2.coef_)

print('Equation: y = ', linear_model2.intercept_, ' + ', linear_model2.coef_[1], 'x + ', linear_model2.coef_[2], 'x^2')
```
