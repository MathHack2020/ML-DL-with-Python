```python
def regression_imputation(target_column):

# Select predictor variables (modify based on correlation results)
   predictors = ['Grade', 'Department_Engineering','Projects_Score',                  
   'Total_Score','Final_Score','Quizzes_Avg' ,'Participation_Score',              
    'Department_CS','Sleep_Hours_per_Night','Stress_Level (1-10)',
    'Study_Hours_per_Week']

# Split data into missing & non-missing attendance
   train_data = df_corr.dropna(subset=[target_column])  # Rows where attendance is available
   missing_data = df_corr[df[target_column].isna()]    # Rows where attendance is missing

# Train regression model
   X_train = train_data[predictors]
   y_train = train_data[target_column]

   regressor = LinearRegression()
   regressor.fit(X_train, y_train)

# Predict missing attendance values
   X_missing = missing_data[predictors]
   df_corr.loc[df_corr[target_column].isna(), target_column] = regressor.predict(X_missing)
```

```python
# Function for random imputation
def random_imputation(column):
    """Replace missing values in a column with random samples from existing values."""
    missing_mask = df_corr[column].isna()  # Identify missing values
    non_missing_values = df_corr[column].dropna().values  # Get non-missing values
    df_corr.loc[missing_mask, column] = np.random.choice(non_missing_values, size=missing_mask.sum(), replace=True)
    return df_corr

# Apply random imputation to Assignments_Avg
df_corr = random_imputation( 'Assignments_Avg')
```

 *  numerical columns also using a Function that using IQR method.

```python
# Function for calculating outliers for Numerical columns
def calculate_num_outliers(df,column):
   Q1 = df[column].quantile(0.25)
   Q3 = df[column].quantile(0.75)
   IQR = Q3-Q1 
   lower_bound = Q1-1.5*IQR
   upper_bound = Q3+1.5*IQR
   outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
   outliers_no =len(outliers)
   if len(outliers)>0 :
       # replace the outlier from the above with the maximum value and from the lower
       # with the lower value.
       df[column] = df[column].clip(lower =lower_bound,upper = upper_bound )
       return outliers_no
```

```python
 # function for handling outlier in categorical column
def calculate_cat_outliers(df, column):

    rare_threshold = 0.05
    frequency = df[column].value_counts(normalize=True)
    outliers = frequency[frequency < rare_threshold].index
    no_of_outliers = len(outliers)

    # If there are some outliers in the categorical columns (categories that are rarely occuring)
    # replace with the mode.

    if outliers.shape[0] > 0:
        df[column] = df[column].replace(outliers, df[column].mode()[0])

    return no_of_outliers
```

```python
# Calculating Outliers in numerical Columns
numeric_outlier_df= pd.DataFrame(columns=['Column','Number of Outliers'])
no_of_numeric_outliers = []
numeric_columns= []
for column in numerical_features :
  outlier_number = calculate_num_outliers(df,column)
  no_of_numeric_outliers.append(outlier_number)
  numeric_columns.append(column)
numeric_outlier_df = pd.DataFrame({'Column':numeric_columns,'Number of Outliers':no_of_numeric_outliers})

numeric_outlier_df.style.background_gradient(cmap='Blues')
```

```python
def plotly_Countplot(features):
  rows = int(np.ceil(len(features)/2))

  fig = make_subplots(rows=rows, cols=2,
                      subplot_titles=[f' Count Plot for {col}' for col in features])
  for i, column in enumerate(features):
    row = i // 2 + 1
    col = i % 2 + 1
    no_of_bins = int(np.ceil(np.sqrt(len(df[column]))))
    if df[column].dtypes == 'category':

       # Define colors for 'Yes' and 'No'
       colors = ['blue', 'red','green','turquoise','violet']  # Choose your desired colors

       for j, grade in enumerate(df['Grade'].unique()):
           fig.add_trace(
              go.Histogram(
                          x=df[df['Grade'] == grade][column],
                          nbinsx=no_of_bins,
                          name=f"{column} - Grade {grade}",
                         # histnorm = 'percent',
                          marker=dict(color=colors[j])  # Use colors list with index j
                          ),
             row=row,
             col=col
                        )

  fig.update_layout(title_text='CountPlot for Categorical Columns',height=row * 300, showlegend=True)

  fig.show(renderer='iframe')
```

# Low Variance Columns

```python
# Low variance in numeric columns
def low_variance_num (df):
  variance =df.var()
  low_variance_columns = variance[variance<0.01].index
  if len(low_variance_columns) >0:
     print(f"Columns with low variance : {low_variance_columns}")
  else :
    print("There is no variance columns")
 # Low variance in categorical columns
def low_variance_cat (df):
  var_cols =[]
  for col in df.columns :
    threshold = 0.9
    counts = df[col].value_counts(normalize=True)

    if counts.max()>threshold :
      var_cols.append(col)

  if len(var_cols)>0:
    print(f"Columns with low variance : {var_cols}")
  else :
    print("There is no variance columns")

```

```python
low_variance_num(df.select_dtypes(include = 'number'))
```

```python
low_variance_cat(df.select_dtypes(include='category'))
```

**Function for returning a coulm of each two hight correlated Columns**

```python
# Function to determine which columns high correlated
# Return the one of two higher correlated coulumns to delete
def correlation (dataset,threshold):
  col_corr = set()
  corr_matrix = dataset.corr()
  for i in range(len(corr_matrix.columns)):
    for j in range(i):
      if abs(corr_matrix.iloc[i,j]) > threshold:
        colname = corr_matrix.columns[i]
        col_corr.add(colname)
  return col_corr
```

```python
col_to_drop=correlation(df[numeric],0.7)
if len(col_to_drop)>0:
  print(f"Columns to drop : {col_to_drop}")
else :
  print('There is no high correlated columns')
```
