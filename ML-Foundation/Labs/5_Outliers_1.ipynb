```python
#   feature normalization using z-score
from scipy import stats
import numpy as np
z=np.abs(stats.zscore(df["1stFlrSF"]))
z
```

```python
# identify the outliers
threshold = 3
print(np.where(z > 3))
```

```python
z[z<3]
```

```python
# we can acomplish teh same and remove the outliers using StandardScaler 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[["1stFlrSF"]])
scaled_data
```

```python
# plot both grahs before and after removing the outliers
# Plotting


feature= "1stFlrSF"
threshold=3
plt.figure(figsize=(12, 5))

# Before Z-score filtering
plt.subplot(1, 3, 1)
plt.hist(df[feature], bins=20, color='blue', alpha=0.7)
plt.title(f'{feature} Before Z-score Filtering')
plt.xlabel('Value')
plt.ylabel('Frequency')

# After Z-score filtering

filtered_data = df[feature][z < threshold]
plt.subplot(1, 3, 2)
plt.hist(z[z<3], bins=20, color='green', alpha=0.7)
plt.title(f'{feature} After Z-score Filtering (Z < 3)')
plt.xlabel('Value')
plt.ylabel('Frequency')



# After scaling
plt.subplot(1, 3, 3)
plt.hist(scaled_data, bins=20, color='orange', alpha=0.7)
plt.title(f'{feature} After StandardScaler')
plt.xlabel('Scaled Value (Z-score)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
